```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return x**2 - 4

x_vals = np.linspace(-6, 6, 400)
y_vals = f(x_vals)


plt.figure(figsize=(8, 5))

plt.plot(x_vals, y_vals, label="y = x^2 - 4")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Quadratic Function")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return x**2 - 4

x_vals = np.linspace(-6, 6, 400)
y_vals = f(x_vals)

x = -5
y = f(x)

plt.figure(figsize=(8, 5))

plt.plot(x_vals, y_vals, label="y = x^2 - 4")

plt.scatter(x, y, color='red', s=50, label="Gradient Descent Step")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Quadratic Function")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return x**2 - 4*x + 1

def grad(x):
    return 2*x - 4

x = np.random.uniform(-10, 10)
lr = 0.1
history = []

for _ in range(50):
    history.append(x)
    x = x - lr * grad(x)

xs = np.linspace(-10, 10, 400)
ys = f(xs)

plt.figure(figsize=(8, 5))
plt.plot(xs, ys, label="y = x^2 - 4x + 1")
plt.scatter(history, f(np.array(history)), color='red', s=30, label="Gradient Descent Steps")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Gradient Descent on Quadratic Function")
plt.legend()
plt.grid(True)
plt.show()
```
