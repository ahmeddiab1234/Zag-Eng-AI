```python
import numpy as np
X = np.c_[[1,1,1],[1,2,3]] 
y = np.c_[[1,3,2]] 
theta = np.linalg.inv(X.T.dot(X)).dot(X.T).dot(y) 
print(theta)
```

```python
X_new = np.c_[[1,1,1,1],[0, 0.5,1.5,4]] 
```

```python
y_pred = X_new.dot(theta)  
print(y_pred)
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([1, 2, 3])
y = np.array([3, 2, 4])

x_line = np.linspace(0, 4, 100)
y_line = 2 + 0.5 * x_line

plt.plot(x_line, y_line, 'r', label=r'$y = 2 + 0.5x_1$')

plt.scatter(x, y, color='black', s=100)

plt.text(1, 3.2, r'$X^{(1)}$', fontsize=12)
plt.text(2, 1.7, r'$X^{(2)}$', fontsize=12)
plt.text(3, 4.2, r'$X^{(3)}$', fontsize=12)

plt.xlabel(r'$x_1$', fontsize=14)
plt.ylabel(r'$y$', fontsize=14, rotation=0, labelpad=10)

plt.xlim(0, 4)
plt.ylim(0, 5)

plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(loc='lower right', fontsize=12)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
X = 2 * np.random.randn(100, 1) 
y = 4 + 35 * X + np.random.randn(100, 1) * 10

X_b = np.c_[np.ones((100, 1)), X] 

theta_best = np.linalg.inv(X_b.T.dot(X_b)).dot(X_b.T).dot(y)

X_new = np.array([[-2], [3]])
X_new_b = np.c_[np.ones((2, 1)), X_new]
y_predict = X_new_b.dot(theta_best)

plt.scatter(X, y, color='blue', label='Data Points')

plt.plot(X_new, y_predict, color='red', linewidth=2, label='Regression Line')

X_pred = np.array([[4]])
X_pred_b = np.c_[np.ones((1, 1)), X_pred]
y_pred = X_pred_b.dot(theta_best)
plt.scatter(X_pred, y_pred, color='green', marker='x', s=100, label='Predictions')

plt.title("Linear Regression using Normal Equation")
plt.xlabel("Feature (X)")
plt.ylabel("Target (y)")
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)

plt.show()
```
